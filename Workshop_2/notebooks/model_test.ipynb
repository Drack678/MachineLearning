```python
import sys
from pathlib import Path
import torch
```

```python
PROJECT_ROOT = Path.cwd().parent
print(PROJECT_ROOT)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

# 2. Importar tus hiperparámetros y la clase de tu modelo
from src.config import VOCAB_SIZE, WINDOW_SIZE, D_MODEL, NHEAD, NUM_LAYERS, DIM_FFW, DROPOUT, BATCH_SIZE
from src.model import ShakespeareTransformer
```

```python
model = ShakespeareTransformer(
    vocab_size=VOCAB_SIZE,
    d_model=D_MODEL,
    nhead=NHEAD,
    num_layers=NUM_LAYERS,
    dim_ffw=DIM_FFW,
    dropout=DROPOUT,
    max_seq_len=WINDOW_SIZE
)
print("¡Modelo construido en memoria!\n")
```

```python
# 4. Crear un lote de datos falsos (Dummy Data)
# Generamos una matriz de tamaño [128, 80] llena de números aleatorios entre 0 y 78 (tus índices del vocabulario)
dummy_x = torch.randint(low=0, high=VOCAB_SIZE, size=(BATCH_SIZE, WINDOW_SIZE))

print(f"-> Tamaño del tensor de ENTRADA: {dummy_x.shape} (Debe ser [128, 80])")

# 5. Prueba de fuego: Pase hacia adelante (Forward Pass)
# Si hay un error matemático en la Atención o el Positional Encoding, el código fallará aquí
output = model(dummy_x)

print(f"-> Tamaño del tensor de SALIDA: {output.shape} (Debe ser [128, 79])")

# 6. Comprobación final
if output.shape == (BATCH_SIZE, VOCAB_SIZE):
    print("\n✅ ¡ÉXITO TOTAL! Tu arquitectura Transformer está perfecta y lista para aprender.")
else:
    print("\n❌ Hay un error en las dimensiones, revisa tu model.py.")
```
